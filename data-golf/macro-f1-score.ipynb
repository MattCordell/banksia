```python
import pandas as pd
from sklearn.metrics import f1_score

def score(solution: pd.DataFrame, submission: pd.DataFrame, row_id_column_name: str) -> float:
    """
    Calculates the macro F1 score across all question columns.
    Each question column is scored independently using binary F1,
    and the results are averaged equally across all questions.
    """
    solution = solution.drop(columns=[row_id_column_name])
    submission = submission.drop(columns=[row_id_column_name])
    
    question_columns = [col for col in solution.columns if col != 'Usage']
    
    f1_scores = []
    for col in question_columns:
        f1 = f1_score(solution[col], submission[col], average='binary')
        f1_scores.append(f1)
    
    return sum(f1_scores) / len(f1_scores)
```
